## Problem  Statement

Student dropout and academic progression are critical challenges in educational institutions. Predicting whether a student will continue studying, drop out, or successfully graduate can help institutions take timely actions.

The aim of this project is to develop a machine learning model that predicts the academic outcome of a student using historical and demographic data. The model classifies students into three categories: Enrolled, Dropout, and Graduate.

# Objective

To analyze student data and identify patterns affecting academic outcomes

To build a multi-class classification model

To predict whether a student will:

1. Continue enrollment

2. Drop out

3. Graduate

## Using Train_test_split

```python
x = student_dataset.drop('Target', axis=1)
y = student_dataset['Target']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size= 0.20, random_state = 42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# Model Training

```python
# Using different models to get better accuracy
```

## 1.Random Forest

```python
rf = RandomForestClassifier()
```

```python
rf.fit(x_train, y_train)
```

```python
y_pred_rf = rf.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_rf)*100
```

## 2. Decision Tree

```python
dt = DecisionTreeClassifier()
```

```python
dt.fit(x_train,y_train)
```

```python
y_pred_dt = dt.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_dt)*100
```

## 3.Logestic Regression

```python
lr = LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred_lr= lr.predict(x_test)
```

```python
accuracy_score(y_test,y_pred_lr)*100
```
